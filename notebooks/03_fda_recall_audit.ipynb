```python
import pandas as pd
import json
import zipfile
```

```python
with zipfile.ZipFile("drug-enforcement-0001-of-0001.json.zip", "r") as z:
    json_file = z.namelist()[0]

    with z.open(json_file) as f:
        recall_data = json.load(f)

recalls = pd.json_normalize(recall_data["results"])

print("Rows:", recalls.shape[0])
print("Columns:", recalls.shape[1])

recalls.head()
```

```python
missing = pd.DataFrame({
    "Missing_Count": recalls.isna().sum(),
    "Missing_Percent":
        (recalls.isna().sum() / len(recalls) * 100).round(2)
})

missing.sort_values(
    "Missing_Percent",
    ascending=False
)
```

```python
date_columns = [
    "recall_initiation_date",
    "report_date",
    "termination_date"
]

for column in date_columns:

    if column in recalls.columns:

        print("\n", column)
        print(recalls[column].head())
```
