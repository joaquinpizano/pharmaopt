```python
import zipfile
import json
import pandas as pd

NDC_ZIP = "drug-ndc-0001-of-0001.json.zip"

with zipfile.ZipFile(NDC_ZIP, "r") as z:
    print(z.namelist())
```

```python
with zipfile.ZipFile(NDC_ZIP, "r") as z:
    with z.open("drug-ndc-0001-of-0001.json") as f:
        ndc_data = json.load(f)

print(ndc_data.keys())
print("Number of NDC records:", len(ndc_data["results"]))
```

```python
ndc = pd.json_normalize(ndc_data["results"])

print("Rows:", ndc.shape[0])
print("Columns:", ndc.shape[1])

ndc.head()
```

```python
ndc.info()
for column in ndc.columns:
    print(column)
```

```python
print("Total columns:", len(ndc.columns))
```

```python
important_columns = [
    "product_ndc",
    "generic_name",
    "brand_name",
    "labeler_name",
    "dosage_form",
    "route",
    "marketing_category",
    "marketing_start_date",
    "marketing_end_date",
    "packaging"
]

existing_columns = [
    col for col in important_columns
    if col in ndc.columns
]

ndc[existing_columns].head(10)
```

```python
package_rows = []

for index, row in ndc.iterrows():

    packages = row.get("packaging")

    if not isinstance(packages, list):
        continue

    for package in packages:

        if not isinstance(package, dict):
            continue

        package_rows.append({
            "product_ndc": row.get("product_ndc"),
            "generic_name": row.get("generic_name"),
            "brand_name": row.get("brand_name"),
            "labeler_name": row.get("labeler_name"),
            "package_ndc": package.get("package_ndc"),
            "package_description": package.get("description")
        })


packages = pd.DataFrame(package_rows)

print("Package records:", len(packages))

packages.head()
```

```python
print("Total package rows:", len(packages))

print(
    "Unique package NDCs:",
    packages["package_ndc"].nunique()
)

print(
    "Missing package NDCs:",
    packages["package_ndc"].isna().sum()
)

print(
    "Duplicate package NDC rows:",
    packages["package_ndc"].duplicated().sum()
)
```

```python
missing = pd.DataFrame({
    "Missing_Count": ndc.isna().sum(),
    "Missing_Percent":
        (ndc.isna().sum() / len(ndc) * 100).round(2)
})

missing.sort_values(
    "Missing_Percent",
    ascending=False
)
```
