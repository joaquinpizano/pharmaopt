```python
import json

with zipfile.ZipFile("drug-shortages-0001-of-0001.json.zip", "r") as z:
    with z.open("drug-shortages-0001-of-0001.json") as f:
        data = json.load(f)

print(data.keys())
```

```python
print("Number of shortage records:", len(data["results"]))
```

```python
data["results"][0]
```

```python
import pandas as pd

shortages = pd.json_normalize(data["results"])

print("Rows:", shortages.shape[0])
print("Columns:", shortages.shape[1])

shortages.head()
```

```python
for column in shortages.columns:
    print(column)
```

```python
shortages.info()
```

```python
shortages["status"].value_counts(dropna=False)
```

```python
shortages["company_name"].value_counts().head(20)
shortages["shortage_reason"].value_counts().head(20)
shortages["therapeutic_category"].value_counts().head(20)
```

```python
missing = pd.DataFrame({
    "Missing_Count": shortages.isna().sum(),
    "Missing_Percent": (shortages.isna().sum() / len(shortages) * 100).round(2)
})

missing.sort_values("Missing_Percent", ascending=False)
```

```python
print("Total records:", len(shortages))

print(
    "Unique package NDCs:",
    shortages["package_ndc"].nunique()
)

print(
    "Missing package NDCs:",
    shortages["package_ndc"].isna().sum()
)
```

```python
date_columns = [
    "initial_posting_date",
    "update_date",
    "change_date"
]

for column in date_columns:
    if column in shortages.columns:
        print(column)
        print(shortages[column].head())
        print()
```
